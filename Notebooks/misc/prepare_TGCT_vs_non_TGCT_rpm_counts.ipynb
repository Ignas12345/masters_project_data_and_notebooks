```python
import pandas as pd
import masters_project_helper_functions.utils as utils
```

```python
url_gtex_rpm_counts = 'https://raw.githubusercontent.com/Ignas12345/masters_project_data_and_notebooks/refs/heads/main/Data/expression_values/GTEx_rpm_counts.csv'
url_tcga_tgct_rpm_counts = 'https://raw.githubusercontent.com/Ignas12345/masters_project_data_and_notebooks/refs/heads/main/Data/expression_values/TCGA_TGCT_mirna_isoform_data_rpm.csv'
url_tcga_non_tgct_rpm_counts = 'https://raw.githubusercontent.com/Ignas12345/masters_project_data_and_notebooks/refs/heads/main/Data/expression_values/TCGA_other_samples_mirna_isoform_data_rpm.csv'
url_TCGA_TGCT_divisions_by_experiment = "https://raw.githubusercontent.com/Ignas12345/masters_project_data_and_notebooks/refs/heads/main/Data/sample_annotations/TCGA_TGCT_divisions_by_experiment.csv"

TCGA_TGCT_divisions_by_experiment = pd.read_csv(url_TCGA_TGCT_divisions_by_experiment, index_col=0)

gtex_mirna_rpm_counts = pd.read_csv(url_gtex_rpm_counts, sep = ';', index_col=0)
tcga_tgct_mirna_rpm_counts = utils.format_data_frame(url_tcga_tgct_rpm_counts, sep = ';', decimal = ',', transpose = True).loc[TCGA_TGCT_divisions_by_experiment.index]
tcga_non_tgct_mirna_rpm_counts = utils.format_data_frame(url_tcga_non_tgct_rpm_counts, sep = ';', decimal = ',', transpose = True)

tcga_tgct_mirna_rpm_counts = utils.initial_pre_processing_pipeline(tcga_tgct_mirna_rpm_counts)
tcga_non_tgct_mirna_rpm_counts = utils.initial_pre_processing_pipeline(tcga_non_tgct_mirna_rpm_counts)
```

```python
#concatenate all 3 data frames in axis 0:
mirna_rpm_counts = pd.concat([tcga_tgct_mirna_rpm_counts, tcga_non_tgct_mirna_rpm_counts, gtex_mirna_rpm_counts], axis=0)
#filter columns that have nan values:
for feature in mirna_rpm_counts.columns:
  if mirna_rpm_counts[feature].isna().sum() > 0:
    mirna_rpm_counts.drop(feature, axis=1, inplace=True)
```

```python
mirna_rpm_counts
```

```python
#save to csv:
mirna_rpm_counts.to_csv('tgct_non_tgct_gtex_combined_mirna_rpm_counts.csv', sep = ';')
```

```python
sample_label_dict = {}
#create a label dict 'TGCT' vs 'non_TGCT':
for sample in mirna_rpm_counts.index:
  if sample in tcga_tgct_mirna_rpm_counts.index:
    sample_label_dict[sample] = 'TGCT'
  else:
    sample_label_dict[sample] = 'non_TGCT'
```

```python
sample_label_dict = pd.DataFrame.from_dict(sample_label_dict, orient='index', columns=['label'])
sample_label_dict.to_csv('TGCT_vs_non_TGCT_labels.csv', sep = ';')
```
