```python
def map_mirnas_to_clusters(input_mirnas, cluster_df):
    """
    Match miRNAs from an input list to their clusters using a reference database.

    Parameters:
    - input_mirnas: list of str, e.g. ["hsa-mir-100, mature,MIMAT0000098"]
    - cluster_df: pd.DataFrame with columns ["name", "Cluster ID"]

    Returns:
    - pd.DataFrame with columns: ["input_miRNA", "base_name", "Cluster ID"]
    """
    # Step 1: Extract base names from input
    base_names = [mirna.split(',')[0].strip() for mirna in input_mirnas]

    # Step 2: Create DataFrame from input list
    input_df = pd.DataFrame({
        "input_miRNA": input_mirnas,
        "base_name": base_names
    })

    # Step 3: Merge with cluster info
    merged_df = input_df.merge(cluster_df, how="left", left_on="base_name", right_on="Name")

    # Optional: Drop redundant 'name' column from cluster_df
    merged_df = merged_df.drop(columns=["Name"])

    return merged_df
```

```python
#calculate feature spearman correlations between features:
#tqdm.pandas()
#corr_matrix = mirna_normalized_isoforms_filtered_df.corr(method='spearman')
corr_matrix = further_filtered_df.corr(method='spearman')
```

```python
threshold = 0.7  # You can tune this
edges = []
weights = []

features = corr_matrix.columns.tolist()

for i in range(len(features)):
    for j in range(i + 1, len(features)):
        weight = corr_matrix.iloc[i, j]
        if abs(weight) >= threshold:
            edges.append((i, j))
            weights.append(abs(weight))  # use abs value as edge weight

# Build igraph object
g = ig.Graph(edges=edges, directed=False)
g.vs["name"] = features
g.es["weight"] = weights
```

```python
def get_features_in_same_module(feature_name, module_df):
    """
    Return all features in the same module as the given feature.

    Parameters:
    - feature_name: str, the feature (e.g. miRNA) of interest
    - module_df: pandas DataFrame with columns ['miRNA', 'module']

    Returns:
    - A list of features in the same module
    """
    if feature_name not in module_df['miRNA'].values:
        print(f"Feature '{feature_name}' not found in the module assignment.")
        return []

    target_module = module_df.loc[module_df['miRNA'] == feature_name, 'module'].values[0]
    features_in_module = module_df.loc[module_df['module'] == target_module, 'miRNA'].tolist()

    print(f"Module {target_module} contains {len(features_in_module)} features:")
    return features_in_module
```
