This notebook is used to download mirna expression profiles for other, non-TGCT, cancer patients. The idea behind using these expression profiles is to detect possible biomarkers that are specific to TGCTs.

```r
if (!requireNamespace("BiocManager", quietly = TRUE))
    install.packages("BiocManager")

BiocManager::install("TCGAbiolinks", ask = FALSE, verbose = TRUE)
```

```r
library(TCGAbiolinks)
library(SummarizedExperiment)
library(dplyr)
library(tidyr)
```

```r
patients <-read.csv2('https://raw.githubusercontent.com/Ignas12345/masters_project_data_and_notebooks/refs/heads/main/Data/sample_annotations/filtered_other_tcga_patients_clin_data.csv')
sumbitter_ids <- patients$submitter_id
```

```r
sumbitter_ids
```

```r
query <- GDCquery(
  project = c("TCGA-ACC", "TCGA-BLCA", "TCGA-BRCA", "TCGA-CESC", "TCGA-CHOL",
              "TCGA-COAD", "TCGA-DLBC", "TCGA-ESCA", "TCGA-GBM", "TCGA-HNSC",
              "TCGA-KICH", "TCGA-KIRC", "TCGA-KIRP", "TCGA-LAML", "TCGA-LGG",
              "TCGA-LIHC", "TCGA-LUAD", "TCGA-LUSC", "TCGA-MESO", "TCGA-OV",
              "TCGA-PAAD", "TCGA-PCPG", "TCGA-PRAD", "TCGA-READ", "TCGA-SARC",
              "TCGA-SKCM", "TCGA-STAD", "TCGA-TGCT", "TCGA-THCA", "TCGA-THYM",
              "TCGA-UCEC", "TCGA-UCS", "TCGA-UVM"),  # You can also use "TCGA" for all
  data.category = "Transcriptome Profiling",
  data.type = "Isoform Expression Quantification",
  barcode = sumbitter_ids
)
```

```r
GDCdownload(query)
```

```r
data <- GDCprepare(query)
```

```r
data_summarized <- data %>%
  mutate(miRNA_id_and_region = paste(miRNA_ID, miRNA_region, sep = ", "))
```

```r
aggregated_data_rpm <- data_summarized %>%
  group_by(miRNA_id_and_region, barcode) %>%
  summarise(total_reads_per_million_miRNA_mapped = sum(reads_per_million_miRNA_mapped), .groups = "drop")
```

```r
final_aggregated_data_rpm <- aggregated_data_rpm %>%
  pivot_wider(names_from = barcode, values_from = total_reads_per_million_miRNA_mapped)
```

```r
head(final_aggregated_data_rpm)
```

```r
write.csv2(final_aggregated_data_rpm, "TCGA_other_samples_mirna_isoform_data_rpm.csv", row.names = FALSE)
```

```r
aggregated_data_raw_counts <- data_summarized %>%
  group_by(miRNA_id_and_region, barcode) %>%
  summarise(total_read_counts = sum(read_count), .groups = "drop")
```

```r
final_aggregated_data_raw_counts <- aggregated_data_raw_counts %>%
  pivot_wider(names_from = barcode, values_from = total_read_counts)
```

```r
head(final_aggregated_data_raw_counts)
```

```r
write.csv2(final_aggregated_data_raw_counts, "TCGA_other_samples_mirna_isoform_data_raw_counts.csv", row.names = FALSE)
```
